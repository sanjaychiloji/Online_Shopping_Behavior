RENAME THE COLUMNS

```python
df.columns = [
    "age_group",
    "gender",
    "occupation",
    "shopping_frequency",
    "device_used",
    "monthly_spending",
    "most_used_platform",
    "product_categories",
    "fashion_platform",
    "electronics_platform",
    "beauty_platform",
    "home_platform",
    "amazon_reason",
    "flipkart_reason",
    "myntra_reason",
    "ajio_reason",
    "meesho_reason",
    "rank_price",
    "rank_quality",
    "rank_reviews",
    "rank_brand",
    "rank_delivery",
    "discount_influence",
    "price_comparison",
    "payment_method",
    "returns_frequency",
    "defective_frequency",
    "overall_experience",
    "delivery_availability",
    "customer_support_rating",
    "problems_faced",
    "trust_new_platform",
    "additional_comments",
    "name",
    "email"
]
```

```python
df.head()
```

CLEAN THE DATA

```python
df = df.drop(columns=["name", "email"])
df
```

```python
df.isnull().sum()
```

```python
df = df.fillna("Not Specified")
```

```python
df.tail(10)
```

```python
#Convert Ranking Columns to Numeric
ranking_cols = [
    "rank_price", "rank_quality",
    "rank_reviews", "rank_brand",
    "rank_delivery"
]

df[ranking_cols] = df[ranking_cols].apply(pd.to_numeric, errors="coerce")
```

```python
for col in ranking_cols:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)
```

```python
df[ranking_cols].isnull().sum()
```

```python
#Convert Rating to Numeric
df["customer_support_rating"] = pd.to_numeric(
    df["customer_support_rating"], errors="coerce"
)
```

ANALYSIS


```python
#Demographic Analysis
age_dist = df["age_group"].value_counts().reset_index()
gender_dist = df["gender"].value_counts().reset_index()
age_dist

```

```python
gender_dist
```

```python
#Platform Usage
platform_usage = df["most_used_platform"].value_counts().reset_index()
platform_usage
```

```python
gender_platform = pd.crosstab(df["gender"], df["most_used_platform"])
gender_platform
```

```python
#Shopping Frequency
shopping_freq = df["shopping_frequency"].value_counts().reset_index()
shopping_freq
```

```python
#Device Usage
device_usage = df["device_used"].value_counts().reset_index()
device_usage
```

```python
#Monthly Spending Analysis
spending_dist = df["monthly_spending"].value_counts().reset_index()
spending_dist
```

```python
#Ranking Analysis
ranking_analysis = df[ranking_cols].mean().sort_values().reset_index()
ranking_analysis.columns = ["Factor", "Average_Rank"]
ranking_analysis
```

```python
#Payment Method
payment_analysis = df["payment_method"].value_counts().reset_index()
payment_analysis
```

```python
#Returns & Defective Products
returns_analysis = df["returns_frequency"].value_counts().reset_index()
defective_analysis = df["defective_frequency"].value_counts().reset_index()
returns_analysis
```

```python
defective_analysis
```

```python
#Overall Experience
experience_analysis = df["overall_experience"].value_counts().reset_index()
experience_analysis
```

```python
#Customer Support Average Rating
avg_support_rating = df["customer_support_rating"].mean()
print("Average Customer Support Rating:", avg_support_rating)
```

VISUALIZATION


```python
plt.figure()
sns.countplot(data=df, x="most_used_platform")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure()
df["age_group"].value_counts().plot(kind="bar")
plt.title("Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("age_distribution.png")
plt.show()
```

```python
plt.figure()
df["gender"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.ylabel("")
plt.tight_layout()
plt.savefig("gender_distribution.png")
plt.show()
```

```python
plt.figure()
df["monthly_spending"].value_counts().plot(kind="bar")
plt.title("Monthly Spending Distribution")
plt.xlabel("Spending Range")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("monthly_spending.png")
plt.show()
```

```python
ranking_cols = [
    "rank_price",
    "rank_quality",
    "rank_reviews",
    "rank_brand",
    "rank_delivery"
]

# Convert to numeric (important)
df[ranking_cols] = df[ranking_cols].apply(pd.to_numeric, errors="coerce")

avg_ranks = df[ranking_cols].mean()

plt.figure()
avg_ranks.plot(kind="bar")
plt.title("Average Ranking of Purchase Factors (Lower = More Important)")
plt.xlabel("Factors")
plt.ylabel("Average Rank")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("ranking_analysis.png")
plt.show()
```

```python
from google.colab import files

all_outputs = {
    "Cleaned_Data": df,
    "Age_Distribution": age_dist,
    "Gender_Distribution": gender_dist,
    "Platform_Usage": platform_usage,
    "Gender_vs_Platform": gender_platform,
    "Shopping_Frequency": shopping_freq,
    "Device_Usage": device_usage,
    "Spending_Distribution": spending_dist,
    "Ranking_Analysis": ranking_analysis,
    "Payment_Methods": payment_analysis,
    "Returns_Analysis": returns_analysis,
    "Defective_Analysis": defective_analysis,
    "Overall_Experience": experience_analysis
}

with pd.ExcelWriter("Final_Online_Shopping_Analysis_.xlsx") as writer:
    for sheet_name, data in all_outputs.items():
        data.to_excel(writer, sheet_name=sheet_name, index=False)

files.download("Final_Online_Shopping_Analysis_.xlsx")

```
